```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

service = Service(ChromeDriverManager().install())
options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BestType=Bestseller&BranchType=1&CID=336"
driver.get(url)
time.sleep(2)

books = driver.find_elements(By.CSS_SELECTOR, "div.ss_book_box")

for i, book in enumerate(books[:55], 1):
    try:
        title_tag = book.find_element(By.CSS_SELECTOR, "a.bo3")
        title = title_tag.text.strip()
    except:
        title = "제목 없음"

    try:
        # 웹 구조 확인 후 CSS -> XPATH 방식으로 변경 전체 li안에 a태그에서 href 속성 값 'AuthorSearch' 포함 확인
        author_links = book.find_elements(By.XPATH, ".//li/a[contains(@href, 'AuthorSearch')]")
        author = author_links[0].text.strip() if author_links else "저자 없음"
    except:
        author = "저자 없음"

    print(f"{i}. {title} - {author}")

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time

# ChromeDriver 서비스 설정
service = Service(ChromeDriverManager().install())

# Chrome 옵션 설정
options = Options()
options.add_argument("--headless")  # 브라우저 화면 없이 실행
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

# 드라이버 실행 (옵션 포함!)
driver = webdriver.Chrome(service=service, options=options)

# 알라딘 베스트셀러 페이지 접속
url = "https://www.aladin.co.kr/shop/common/wbest.aspx?BestType=Bestseller&BranchType=1&CID=336"
driver.get(url)

# 페이지 로딩 대기
time.sleep(2)

# 페이지 전체 HTML 가져오기
soup = BeautifulSoup(driver.page_source, "html.parser")

# 책 리스트 찾기
book_items = soup.select("div.ss_book_box")

# 상위 10개만 출력
for i, book in enumerate(book_items[:10], 1):
    title_tag = book.select_one("a.bo3")
    author_tag = book.select_one("div.ss_book_list ul li:nth-child(3) a")  # 저자 정보는 3번째 li 안에 있음

    title = title_tag.text.strip() if title_tag else "제목 없음"
    author = author_tag.text.strip() if author_tag else "저자 없음"

    print(f"{i}. {title} - {author}")

# 브라우저 종료
driver.quit()
```
