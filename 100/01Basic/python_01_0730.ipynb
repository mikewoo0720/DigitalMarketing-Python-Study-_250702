#### 6. 잡코리아에서 '데이터 분석가' 직무 공고 제목 20개를 수집하세요. 
- 셀레니움 코드로 크롤링 하기
- 6개 페이지 크롤링 해오기
- 엑셀로 출력 (*해당 공고 타이틀과 지원할 수 있는 URL 링크)
- 서울지역 내 직무 공고 & 경력 1~3년차만 엑셀 1열 타이틀 2열 링크

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from openpyxl import Workbook
import time
from urllib.parse import quote

# 엑셀파일
wb = Workbook()
ws = wb.active
ws.title = "Jobkorea_info"
ws.append(["공고 제목", "링크"])

# 크롬 드라이버 설정
service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")

options = Options()
options.add_argument("--headless")  # 창 숨기기
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=service, options=options)

# 검색어 설정
query = "데이터 분석가"
encoded_query = quote(query)  # 한글 인코딩
base_url = f"https://www.jobkorea.co.kr/Search?stext={encoded_query}"

# 여러 페이지 탐색
for n in range(1, 7):
    print(f"\n--- Page {n} ---")
    url = f"{base_url}&Page_No={n}&local=1000"
    driver.get(url)
    time.sleep(2)  # 페이지 로딩 대기

    job_elements = driver.find_elements(By.CSS_SELECTOR, "div.styles_mb_space8__dk46ts1p a.sn28bt0")
    
    for job in job_elements:
        title = job.text.strip()
        link = job.get_attribute("href")
        ws.append([title, link])
        print(f"{title}\n→ {link}")

driver.quit()

# 엑셀 저장
wb.save("Jobkorea_info.xlsx")
print("✅ openpyxl로 엑셀 저장 완료!")
```

```python
#선생님
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
import urllib.parse
import pandas as pd

service = Service(ChromeDriverManager().install())

options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = "데이터 분석가"
encoded_query = urllib.parse.quote(query)

max_pages = 6

results = []

for page in range(1, max_pages + 1) :
    url = f"https://www.jobkorea.co.kr/Search?stext={encoded_query}&Page_No={page}&local=1000&tabType=recruit&careerType=1"
    driver.get(url)
    time.sleep(2)

    job_posts = driver.find_elements(By.CSS_SELECTOR, "div.styles_mb_space8__dk46ts1p")
    for post in job_posts :
        try :
            title_elem = post.find_element(By.CSS_SELECTOR, "a.sn28bt0")
            title = title_elem.text.strip()
            link = title_elem.get_attribute("href")
            results.append({
                "제목" : title,
                "지원링크" : link
            })
        except :
            continue

driver.quit()

print(results)

df = pd.DataFrame(results)
df.to_excel("잡코리아_데이터분석가_서울_신입_3페이지.xlsx", index=False)
print("엑셀 저장 완료!")
```
