```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
import re

# 드라이버 설정
service = Service(ChromeDriverManager().install())
options = Options()
options.add_argument('--window-size=1920x1080')
options.add_argument('--start-maximized')
options.add_argument('--lang=ko_KR')

driver = webdriver.Chrome(service=service, options=options)
driver.get("https://www.musinsa.com/main/musinsa/ranking")
time.sleep(3)

items = []

# 10번 정도 스크롤 → 50개 확보 가능
for _ in range(10):
    ranks = driver.find_elements(By.CSS_SELECTOR, "div.sc-1t5ihy5-9.iXBmPD")

    for rank in ranks[len(items):]:  # 아직 안 저장한 상품만
        try:
            brand = rank.find_element(By.CSS_SELECTOR, 
                "p.text-etc_11px_semibold.line-clamp-1.break-all.whitespace-break-spaces.text-black.font-pretendard").text
            title = rank.find_element(By.CSS_SELECTOR, 
                "p.text-body_13px_reg.line-clamp-2.break-all.whitespace-break-spaces.text-black.font-pretendard").text
            price = rank.find_element(By.CSS_SELECTOR, 
                "span.text-body_13px_semi.sc-1t5ihy5-11.cWXWMC.text-black.font-pretendard").text
            try:
                dis_percent = rank.find_element(By.CSS_SELECTOR, 
                    "span.text-body_13px_semi.sc-1t5ihy5-11.cWXWMC.text-red.font-pretendard").text
            except:
                dis_percent = "0%"

            # 상품 상세 URL
            product_link = rank.find_element(By.CSS_SELECTOR, "a.gtm-click-detail").get_attribute("href")

            # 새 탭 열어서 리뷰수 크롤링
            driver.execute_script("window.open(arguments[0]);", product_link)
            driver.switch_to.window(driver.window_handles[1])
            time.sleep(2)

            try:
                review_element = driver.find_element(By.CSS_SELECTOR, 
                    "a.style-review-btn strong")  # 리뷰수 표시 요소
                review_count = int(re.sub(r"[^\d]", "", review_element.text))
            except:
                review_count = 0

            # 탭 닫고 원래 페이지로
            driver.close()
            driver.switch_to.window(driver.window_handles[0])

            items.append((brand, title, price, dis_percent, review_count))
        except:
            pass

        if len(items) >= 50:
            break

    if len(items) >= 50:
        break

    driver.execute_script("window.scrollBy(0, 1000);")
    time.sleep(2)

driver.quit()

# 데이터프레임 저장
df = pd.DataFrame(items, columns=["브랜드", "제품명", "가격", "할인율", "리뷰수"])
print(df)

df.to_csv("musinsa_ranking_with_reviews.csv", index=False, encoding="utf-8-sig")
```
