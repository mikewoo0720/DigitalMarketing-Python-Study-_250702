```python
#4. Github 로그인 후 내 레포지토리 리스트를 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import pickle
import time

# 옵션 설정
options = Options()
prefs = {
    "profile.default_content_setting_values.notifications": 1  # 1:허용, 2:차단
}
options.add_experimental_option("prefs", prefs)
options.add_argument("--start-maximized")
options.add_argument("window-size=1920x1080")
options.add_argument("lang=ko_KR")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

# 2. 쿠키 불러오기

driver.get("https://github.com/")  # 도메인 먼저 접속

with open("github_cookies.pkl", "rb") as f:
    cookies = pickle.load(f)

for cookie in cookies:
    if "github.com" not in cookie.get("domain", ""):
        continue
    if cookie.get("httpOnly", False):
        continue
    if cookie.get("sameSite") == "None":
        cookie["secure"] = True
    if "expiry" in cookie and not isinstance(cookie["expiry"], int):
        cookie.pop("expiry")
    try:
        driver.add_cookie(cookie)
    except Exception as e:
        print(f"쿠키 추가 실패: {cookie.get('name')}, 사유: {e}")

driver.refresh()

time.sleep(3)

target_page_url = "https://github.com/mikewoo0720/DigitalMarketing-Python-Study-_250702"
driver.get(target_page_url)

time.sleep(3)  # 페이지 로딩 대기


titles = driver.find_elements(By.CSS_SELECTOR, "tr.react-directory-row.undefined")

for i, title in enumerate(titles, 1) :
    print(f"{i}. {title.text} \n")

driver.quit()
```
