```python
#마켓컬리에서 신상품 중 평점이 높은 10개 상품의 이름과 평점을 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib
import time

service = Service(ChromeDriverManager().install())

options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = ""
encoded_query = urllib.parse.quote(query)

driver.get(f"https://www.kurly.com/collection-groups/market-newproduct?page=1&collection=newranking1")
time.sleep(1)

datas = driver.find_elements(By.CSS_SELECTOR, "div.css-1kpzrna.e1c07x486")

for i, data in enumerate(datas[:10], 1):
    try:
        title = data.find_element(By.CSS_SELECTOR, "span.css-1dry2r1.e1c07x485").text
    except:
        title = "제목 없음"

    try:
        review = data.find_element(By.CSS_SELECTOR, "span.review-number").text
    except:
        review = "0"

    print(f"{i}. 제목: {title} | 리뷰 수: {review}")

driver.quit()
```

```python
#G마켓에서 '게이밍 키보드' 검색 후 스크롤을 내려 상품명과 리뷰수를 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib
import time

service = Service(ChromeDriverManager().install())

options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

keyword = "게이밍 키보드"
encoded_keyword = urllib.parse.quote(keyword)
driver.get(f"https://browse.gmarket.co.kr/search?keyword={encoded_keyword}")
time.sleep(2)

collected = []
titles = []
reviews = []

for _ in range(10):  # 최대 10번 스크롤
    items = driver.find_elements(By.CSS_SELECTOR, "div.box__component-itemcard")
    
    for item in items:
        try:
            title = item.find_element(By.CSS_SELECTOR, "span.text__item").text.strip()
        except:
            title = "제목 없음"

        try:
            review = item.find_element(By.CSS_SELECTOR, "li.list-item.list-item__feedback-count > span.text").text.strip()
        except:
            review = "0"

        key = f"{title}__{review}"
        if key not in collected:
            collected.append(key)
            titles.append(title)
            reviews.append(review)

        if len(titles) >= 100:
            break

    if len(titles) >= 100:
        break

    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(1.2)

for i in range(len(titles)):
    print(f"{i+1}. 상품명: {titles[i]} / 리뷰 수: {reviews[i]}")

driver.quit()
```

```python
#11번가에서 '캠핑용품' 검색 후 가격 범위 필터 적용 후 상품명 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib
import time

service = Service(ChromeDriverManager().install())

options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = "pc/total-search?kwd=%25EC%25BA%25A0%25ED%2595%2591%25EC%259A%25A9%25ED%2592%2588&tabId=TOTAL_SEARCH"
driver.get(f"https://search.11st.co.kr/{query}")
time.sleep(2)

titles = driver.find_elements(By.CSS_SELECTOR, "div.c-card-item__name")
prices = driver.find_elements(By.CSS_SELECTOR, "dd.c-card-item__price")

print("가격이 30,000원 이상인 상품:")
count = 1
for title, price in zip(titles, prices):
    try:
        price_text = price.text.replace(",", "").replace("원", "").strip()
        price_value = int(price_text)

        if price_value >= 30000:
            print(f"{count}. {title.text.replace("\n", " : ")} / 가격: {price.text}")
            count += 1
    except:
        continue #둘다 없을 경우 무시

driver.quit()
```

```python
#GS샵에서 홈쇼핑 생방송 시간표를 스크롤로 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib
import time

service = Service(ChromeDriverManager().install())

options = Options()
# options.add_argument("--headless")  
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

# 크롤링 원하는 날짜 수집할 수 있도록 쿼리값 설정 
query = "#20250808"
driver.get(f"https://www.gsshop.com/shop/tv/tvScheduleMain.gs?lseq=415680-1&gsid=ECmain-AU415680-AU415680-1{query}_LIVE")
time.sleep(2)


for _ in range(10):
    items = driver.find_elements(By.CSS_SELECTOR, "article.items")
    for item in items:
        try:
            title = item.find_element(By.CSS_SELECTOR, "dt.prd-name").text.strip()
        except:
            title = "제목 없음"
        try:
            time_info = item.find_element(By.CSS_SELECTOR, "span.times").text.strip()
        except:
            time_info = "시간 없음"
        print(f"시간: {time_info} / 제목: {title}")
            
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(1.5)

driver.quit()
```
