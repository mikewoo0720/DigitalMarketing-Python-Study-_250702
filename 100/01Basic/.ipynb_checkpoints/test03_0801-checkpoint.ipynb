```python
#노컷뉴스 연예면 주요 기사 제목을 5개 수집하세요.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib
import time

service = Service(ChromeDriverManager().install())

options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = "news/entertainment"
encoded_query = urllib.parse.quote(query)

url = f"https://www.nocutnews.co.kr//{query}"
driver.get(url)

texts = driver.find_elements(By.CSS_SELECTOR, "strong.clamp2")

for i, text in enumerate(texts[:5]) :
    print(f"{i+1} : {text.text}")


driver.quit()
```

```python
#아프리카TV 인기 방송 제목을 수집하세요.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib
import time

service = Service(ChromeDriverManager().install())

options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = ""
encoded_query = urllib.parse.quote(query)

url = f"https://www.sooplive.co.kr/{query}"
driver.get(url)
time.sleep(1)

texts = driver.find_elements(By.CSS_SELECTOR, "h3.title")

for i, text in enumerate(texts[:10]) :
    print(f"{i+1} : {text.text}")


driver.quit()
```

```python
#마이리얼트립에서 인기 도시별 여행 상품 제목을 수집하세요.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
import urllib.parse

service = Service(ChromeDriverManager().install())

options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = "" 
encoded_query = urllib.parse.quote(query)
url = f"https://www.myrealtrip.com/{query}"
driver.get(url)
time.sleep(1)


fav_button = driver.find_element(By.CSS_SELECTOR, "div.css-1f9jvpk")
fav_button.click()
time.sleep(3)

for i in range(1, 8):
    xpath = f"(//div/button)[{i}]"
    
    try:
        print(f"{i}번째 버튼 클릭 시도 중...")
        button = driver.find_element(By.XPATH, xpath)
        button.click()
        time.sleep(2)
        
        #다시 열기 필요 시 주석 해제
        # fav_button = driver.find_element(By.CSS_SELECTOR, "div.css-1f9jvpk")
        # fav_button.click()
        # time.sleep(2)

    except Exception as e:
        print(f"{i}번째 버튼 클릭 실패: {e}")

driver.quit()
```
