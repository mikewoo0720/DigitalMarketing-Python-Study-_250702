```python
#위키백과에서 '마케팅' 문서의 첫 문단 요약을 추출하세요.

from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

service = Service(ChromeDriverManager().install())

options = Options()
options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = "마케팅"
encoded_query = urllib.parse.quote(query)

url = f"https://ko.wikipedia.org/wiki/{query}"
driver.get(url)
time.sleep(1)

title_lists = driver.find_elements(By.CSS_SELECTOR, "dl.type-b > dt")

for i, title in enumerate(title_lists[:10]) :
    print(f"{i+1} : {title.text}")


driver.quit()
```

```python
#인터파크 티켓에서 인기 콘서트 제목과 일정을 수집하세요.
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as ec
from webdriver_manager.chrome import ChromeDriverManager
import time

options = Options()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--start-maximized")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--lang=ko_KR")

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service = service  , options = options)

query = "?genre=CONCERT"
encoded_query = urllib.parse.quote(query)

url = f"https://tickets.interpark.com/contents/ranking{query}"
driver.get(url)
time.sleep(1)

title_list = driver.find_elements(By.XPATH , "//*[contains (@class, 'responsive-ranking-list_rankingContents__')]")

for i, title_list in enumerate(title_list,1) :
    title = title_list.find_elements(By.XPATH, ".//li")[0].text
    date = title_list.find_elements(By.XPATH, ".//li")[2].text
    print(f"{i}: 제목 : {title} 날짜 :{date}\n")

time.sleep(1)

driver.quit()
```

```python
#삼성전자 뉴스룸에서 최근 기사 제목과 날짜를 수집하세요.최종
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import urllib.parse
import time

service = Service(ChromeDriverManager().install())

options = Options()
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)

query = ""
encoded_query = urllib.parse.quote(query)
url = f"https://news.samsung.com/kr/{query}"
driver.get(url)

time.sleep(2)


title_lists = driver.find_elements(By.CSS_SELECTOR, "ul.item > li")

for i, title_list in enumerate(title_lists[:10]):
    try:
        title = title_list.find_element(By.CSS_SELECTOR, "span.ellipsis")
        date = title_list.find_element(By.CSS_SELECTOR, "span.date")
        print(f"{i+1}. 제목 : {title.text.strip()} | 날짜 : {date.text.strip()}")
    except:
        print(f"{i+1}. 없습니다.")

time.sleep(2)
driver.quit()
```
