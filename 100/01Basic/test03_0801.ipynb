```python
#마이리얼트립에서 인기 도시별 여행 상품 제목을 수집하세요.v2 수정
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from webdriver_manager.chrome import ChromeDriverManager
import time
import urllib.parse

# 드라이버 설정
service = Service(ChromeDriverManager().install())
options = Options()
# options.add_argument("--headless")
options.add_argument("--window-size=1920x1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)
actions = ActionChains(driver)

# 페이지 열기
url = "https://www.myrealtrip.com/"
driver.get(url)
time.sleep(2)

# 인기 도시 보기 클릭
fav_button = driver.find_element(By.CSS_SELECTOR, "div.css-1f9jvpk")
fav_button.click()
time.sleep(2)

# 도시 목록 불러오기
buttons_city = driver.find_elements(By.CSS_SELECTOR, "button.css-8nly55")

# 각 도시 클릭
for city_index in range(len(buttons_city)):
    buttons_city = driver.find_elements(By.CSS_SELECTOR, "button.css-8nly55")
    button = buttons_city[city_index]
    city_name = button.text.strip()
    print(f"\n▶ 도시 {city_index}: {city_name}")
    
    button.click()
    time.sleep(2)

    # 인기 도시 다시 열기
    fav_button.click()
    time.sleep(2)

    # 드래그 대상 영역 (수평 슬라이드 영역)
    try:
        drag_area = driver.find_element(By.CSS_SELECTOR, "section.css-11zhy3w")
    except:
        print("드래그 영역 없음")
        continue

    # 여러 번 오른쪽으로 드래그
    for _ in range(5):  # 5번 드래그 시도
        actions.click_and_hold(drag_area)
        actions.move_by_offset(-700, 0)  # 오른쪽으로 스와이프
        actions.release()
        actions.perform()
        time.sleep(1)

    # 여행 상품 제목 수집
    titles = driver.find_elements(By.CSS_SELECTOR, "span.css-43qqj2")

    for i, title in enumerate(titles):
        text = title.text.strip()
        if text:
            print(f"{i}. {text}")
        else:
            print(f"{i}. [빈 항목]")

# 종료
driver.quit()
```
