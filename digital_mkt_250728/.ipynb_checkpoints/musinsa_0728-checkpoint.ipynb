```python
#무신사 리바이스 리뷰 제일 많은 내용 크롤링
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait#천천히 찾을 수 있도록 대기시켜줄 수 있는 프로토타입 객체
from selenium.webdriver.support import expected_conditions as EC#클릭할 수 있는 요소인지 아닌지 판단할 수 있도록 도와주는 프로토타입 객체
from bs4 import BeautifulSoup
from webdriver_manager.chrome import ChromeDriverManager#웹 버전 맞추기
import time
from openpyxl import Workbook

options = Options()
# options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")#로컬 공간에서 메모리 사용할 수 있도록 설정
options.add_argument("--start-maximized")
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

url = "https://www.musinsa.com/products/2439102"
driver.get(url)

time.sleep(2)
try:
    popup_close_btn = WebDriverWait(driver, 3).until(
        EC.element_to_be_clickable((By.CSS_SELECTOR, "div.ab-in-app-message.ab-background.ab-modal-interactions.ab-modal.ab-centered > button"))
    )
    time.sleep(1)
    popup_close_btn.click()
    print("팝업 닫기 완료")
    time.sleep(1)
except:
    print("팝업 없음, 계속 진행")

#try-except문 트라이가 실행이 안되면 익셉트가 실행된다. 이 구문을 쓸 경우 어디서 오류났는지 빠르게 찾을 수 있는 장점이 있음
try :#(자동화가 되어있음) 10초 기다리는 동안 버튼은 아니지만 버튼의 기능을 할 수 있는지 찾아줘
    review_tab = WebDriverWait(driver, 10).until(
        EC.element_to_be_clickable((By.XPATH, "//span[contains(text(), '스냅·후기')]"))#위치는 바뀔 수 있어도 속성값은 변하지 않으니 XPATH 방식으로 찾는 것이 유리
    )
    review_tab.click()
    time.sleep(2)
except Exception as e :
    print("리뷰 탭 클릭 실패 : ", e)

scroll_pause = 1
last_height = driver.execute_script("return document.body.scrollHeight")

for _ in range(5) :
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")
    time.sleep(scroll_pause)
    new_height = driver.execute_script("return document.body.scrollHeight")
    if new_height == last_height :
        break
    last_height = new_height

soup = BeautifulSoup(driver.page_source, "html.parser")

product_name = soup.select_one("span.text-title_18px_med.sc-1omefes-1.exqQRL.font-pretendard").get_text(strip=True)

review_cards = soup.select("div.GoodsReviewStaticList__Container-sc-s1ucwv-0.lnqPxe")

reviews = []
for card in review_cards :
    text_tags = card.select(".text-body_13px_reg.text-black.font-pretendard")
    for tag in text_tags :
        text = tag.get_text(strip=True)
        if text and len(text) > 10 :
            reviews.append(text)

reviews = list(set(reviews))#set 함수를 통해서 중복된 내용 제거

print(f"상품명 : {product_name}")
for idx, review in enumerate(reviews, start=1) :
    print(f"{idx}. {review}")
    
driver.quit()
```
