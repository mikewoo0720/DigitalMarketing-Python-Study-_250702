```python
#이때까지 배운 내용 종합해서 컬리 페이지 특정상품 리뷰 엑셀로 저장
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from openpyxl import Workbook
import time

#엑셀파일
wb = Workbook()
ws = wb.active
ws.title = "Kurly Data"
ws.append(["text"])

# 크롬 드라이버 설정
service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")

options = Options()
# options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
options.add_argument("--disable-save-password-bubble")
options.add_argument("--disable-infobars")
options.add_argument("--disable-extensions")

driver = webdriver.Chrome(service=service, options=options)

url = "https://www.kurly.com/goods/1000130632"
driver.get(url)

# next_buttons = driver.find_elements(By.XPATH, "//button[@class = 'css-1orps7k ebs5rpx0']")
# print(next_buttons)

max_pages = 200  # 안전을 위해 최대 페이지 수 지정

for n in range(1, max_pages + 1):
    print(f"\n--- Page {n} ---")
    
    elements = driver.find_elements(By.XPATH, "//div[@class = 'css-169773r e36z05c8']")
    for element in elements:
        ws.append([element.text])
        print(element.text)

    # 다음 페이지 버튼이 있는지 체크
    next_buttons = driver.find_elements(By.XPATH, "//button[@class = 'css-1orps7k ebs5rpx0']")
    if len(next_buttons) == 0:
        print("더 이상 다음 페이지가 없습니다. 크롤링 종료.")
        break
    
    # 다음 페이지 클릭
    element_nextpg = next_buttons[0]
    driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", element_nextpg)
    time.sleep(1)
    element_nextpg.click()
    time.sleep(2)

driver.quit()

wb.save("kurly_crawled_data_openpyxl.xlsx")

print("openpyxl로 엑셀 저장 완료!")
```

```python
#번개장터 크롤링 코드 챗pgt
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import traceback

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")

options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service=service, options=options)

try:
    url = "https://m.bunjang.co.kr/shop/77183378/reviews?"
    driver.get(url)

    wait = WebDriverWait(driver, 10)

    # 텍스트가 20자 이상인 모든 div를 리뷰로 간주
    elements = wait.until(
        EC.presence_of_all_elements_located((By.XPATH, "//div[string-length(text()) > 20]"))
    )

    for idx, element in enumerate(elements, 1):
        print(f"[{idx}] {element.text.strip()}")

except Exception as e:
    print("❌ 오류 발생:", e)
    traceback.print_exc()

finally:
    driver.quit()
```
