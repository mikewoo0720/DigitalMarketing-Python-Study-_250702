#### HTML 크롤링을 위한 XPATH 기본 포맷 (*형태)
- //tagname[@Attribute="value"] => 기본 포맷
- / : 절대경로를 의미 : (ex. /html/body/div/a)
- // : 상대경로 : 문서 내에서 바로 검색을 하라는 의미 (ex. //h1 -> 문서에서 바로 h1 태그를 찾아와라)
- //*[@href] : 문서 내 href라는 속성을 가지고 있는 모든 태그를 다 찾아와라
- //a[@href="https//www.google.com"] : 문서 내 a태그를 찾아오는데, href 속성값이 구글 닷컴 인 데이터를 찾아와라
- (//a)[3] : 문서 내 a태그 중 3번째 a태그 데이터만 찾아와라 (주의 : 문서상에서 '1'부터 카운트한다)
- (//table)[last] : 문서 내 table태그 중 마지막 데이터만 찾아와라
- (//a)[position() < 3] : 문서 내 a 태그를 찾아오는데, a 태그 데이터 중 앞에서부터 2개만 찾아와라
- //div[@*] : 문서 내 div 태그를 찾아오는데, 어떤 속성이든 상관없이 속성이 하나라도 있으면 찾아와라(@은 속성값을 의미)
- //*[contains(@class, '')] : class 전용 함수
아주 날카로운 질문이에요.
결론부터 말하자면:

✅ class='값'도 여전히 필요할 때가 있습니다.
단, 상황에 따라 contains()가 더 유연하거나, class='값'이 더 정확합니다.

✅ 차이점 정리
XPath 표현식	설명	정확도	유용성
@class='course'	클래스가 정확히 'course' 하나만 있는 경우	🎯 매우 정확	제한적
contains(@class, 'course')	클래스에 'course'가 포함만 되어 있으면 OK (다른 클래스들과 함께 있어도 됨)
