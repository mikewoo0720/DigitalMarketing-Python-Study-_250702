```python
#이때까지 배운 내용 종합해서 컬리 페이지 특정상품 리뷰 엑셀로 저장
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from openpyxl import Workbook
import time

#엑셀파일
wb = Workbook()
ws = wb.active
ws.title = "Kurly Data"
ws.append(["text"])

# 크롬 드라이버 설정
service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")

options = Options()
# options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
options.add_argument("--disable-save-password-bubble")
options.add_argument("--disable-infobars")
options.add_argument("--disable-extensions")

driver = webdriver.Chrome(service=service, options=options)

url = "https://www.kurly.com/goods/1000130632"
driver.get(url)

# next_buttons = driver.find_elements(By.XPATH, "//button[@class = 'css-1orps7k ebs5rpx0']")
# print(next_buttons)

max_pages = 200  # 안전을 위해 최대 페이지 수 지정

for n in range(1, max_pages + 1):
    print(f"\n--- Page {n} ---")
    
    elements = driver.find_elements(By.XPATH, "//div[@class = 'css-169773r e36z05c8']")
    for element in elements:
        ws.append([element.text])
        print(element.text)

    # 다음 페이지 버튼이 있는지 체크
    next_buttons = driver.find_elements(By.XPATH, "//button[@class = 'css-1orps7k ebs5rpx0']")
    if len(next_buttons) == 0:
        print("더 이상 다음 페이지가 없습니다. 크롤링 종료.")
        break
    
    # 다음 페이지 클릭
    element_nextpg = next_buttons[0]
    driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", element_nextpg)
    time.sleep(1)
    element_nextpg.click()
    time.sleep(2)

driver.quit()

wb.save("kurly_crawled_data_openpyxl.xlsx")

print("openpyxl로 엑셀 저장 완료!")
```

```python
#번개장터 크롤링 코드 챗pgt => xpath로 현이네 상점 한페이 분량(두개)만큼 데이터 가져오기 성공
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import traceback

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")

options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service=service, options=options)

try:
    url = "https://m.bunjang.co.kr/shop/77183378/reviews?"
    driver.get(url)

    wait = WebDriverWait(driver, 10)

    elements = wait.until(
        EC.presence_of_all_elements_located((By.XPATH, "//*[@id='root']/div/div/div[4]/div/div[5]/div/div[2]"))
    )

    for idx, element in enumerate(elements, 1):
        print(f"[{idx}] {element.text.strip()}")

except Exception as e:
    print("❌ 오류 발생:", e)
    traceback.print_exc()

finally:
    driver.quit()
```

```python
#상점후기 더보기 마지막 까지 클릭 후 후기내용 가져오기
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException, NoSuchElementException
import time
import traceback

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service=service, options=options)

try:
    url = "https://m.bunjang.co.kr/shop/77183378/reviews?"
    driver.get(url)
    wait = WebDriverWait(driver, 10)

    # 1. 🔁 모든 "더보기" 버튼 클릭해서 리뷰 전체 로딩
    while True:
        try:
            more_button = wait.until(EC.element_to_be_clickable(
                (By.XPATH, '//*[@id="root"]/div/div/div[4]/div/div[5]/div/div[3]/button')))
            driver.execute_script("arguments[0].click();", more_button)
            time.sleep(1.2)  # 클릭 후 로딩 대기
        except (TimeoutException, NoSuchElementException):
            print("✅ 더 이상 '더보기' 버튼이 없습니다.")
            break

    time.sleep(1.5)  # 마지막 리뷰 로딩 기다림

    # 2. ✂️ 사용자 리뷰 텍스트 추출 (정확한 XPath 사용)
    review_elements = driver.find_elements(By.XPATH, "//*[@id='root']/div/div/div[4]/div/div[5]/div/div[2]")

    # 3. 🔍 불필요한 텍스트 필터링
    reviews = []
    for element in review_elements:
        text = element.text.strip()
        if not text:
            continue
        # 사업자정보나 안내문 제외 키워드 필터
        if any(kw in text for kw in [
            "대표이사", "사업자등록번호", "환불", "반품", "교환", "공지사항", "All rights reserved",
            "센터필드", "EMAIL", "FAX", "Amazon", "운영시간", "고객", "소비자보호", "회사가 직접"
        ]):
            continue
        reviews.append(text)

    # 4. 🖨 리뷰 출력
    print(f"\n💬 총 {len(reviews)}개의 사용자 리뷰를 수집했습니다.\n")
    for idx, review in enumerate(reviews, 1):
        print(f"[{idx}] {review}")

except Exception as e:
    print("❌ 오류 발생:", e)
    traceback.print_exc()

finally:
    driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import pandas as pd
import time
import pyautogui
# :흰색_확인_표시: ChromeDriver 설정
service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=service, options=options)
# :흰색_확인_표시: URL
url = "https://m.bunjang.co.kr/shop/3541801/reviews?"
driver.get(url)
# :흰색_확인_표시: 스크롤 & 알림 닫기 반복
for i in range(100):  # 필요시 횟수 늘려주세요
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(0.3)
    pyautogui.click(x=930, y=470)  # 버튼 좌표
    time.sleep(0.3)
time.sleep(2)
# :흰색_확인_표시: 개별 리뷰 블록
review_blocks = driver.find_elements(By.XPATH, "//*[@id='root']/div/div/div[4]/div/div[5]/div/div[2]/div")
data = []
keyword_patterns = ["상품", "배송", "번개톡", "포장"]

for block in review_blocks:
    lines = block.text.strip().split("\n")
    if len(lines) < 3:
        continue
    nickname = lines[0].strip()
    date = lines[1].strip()
    product = lines[2].strip()
    # :흰색_확인_표시: 리뷰 텍스트는 상품명 바로 아래 줄만 확인
    review_text = "None"
    if len(lines) > 3:
        candidate = lines[3].strip()
        if not any(k in candidate for k in keyword_patterns) and "+" not in candidate:
            review_text = candidate
    # :흰색_확인_표시: 키워드 수집 (4번째 줄 이후)
    keywords = []
    for line in lines[3:]:
        if any(k in line for k in keyword_patterns):
            keywords.append(line)
    keyword_text = ", ".join(keywords) if keywords else "None"
    data.append({
        "닉네임": nickname,
        "날짜": date,
        "상품명": product,
        "리뷰내용": review_text,
        "키워드": keyword_text
    })
driver.quit()
# :흰색_확인_표시: 엑셀 저장
df = pd.DataFrame(data)
df.to_excel("bunyang_reviews_fixed.xlsx", index=False)
print(":흰색_확인_표시: 크롤링 완료 → bunjang_reviews.xlsx 저장됨!")
```
