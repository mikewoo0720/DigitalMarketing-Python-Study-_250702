#### 6. 다음 웹사이트에 접속해서 selenium을 활용해 크롤링 해오세요.
- https://divjason.github.io/sellenium-test/
- 보여지는 input tag 요소에 id 값으로 error@error.com password 값으로 1234 입력 후 로그인 버튼 자동으로 클릭하기
- 클릭 이후 이동한 페이지에서 message 라는 클래스 속성겂을 가지고 있는 요소의 텍스트 정보를 출력!!

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time
import pyautogui

from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service = service, options = options)

url = "https://divjason.github.io/sellenium-test/"
driver.get(url)

element = driver.find_element(By.NAME, "username")
element.clear()
element.send_keys("error@error.com")

element = driver.find_element(By.NAME, "password")
element.clear()
element.send_keys("1234")

# element.send_keys(Keys.RETURN)
element = driver.find_element(By.CSS_SELECTOR, "input[type='submit']")

element.click()

elements = driver.find_elements(By.CSS_SELECTOR, "div.message")

time.sleep(3)

for element in elements :
    print(element.text)

time.sleep(3)

# driver.save_screenshot("shot1.png")

driver.quit()
```

#### 6. 다음 웹사이트에 접속해서 selenium을 활용해 크롤링 해오세요.
- https://divjason.github.io/sellenium-test/
- 보여지는 input tag 요소에 id 값으로 error@error.com password 값으로 1234 입력 후 로그인 버튼 자동으로 클릭하기
- 클릭 이후 이동한 페이지에서 message 라는 클래스 속성겂을 가지고 있는 요소의 텍스트 정보를 출력!!
- 해당 페이지에서 로그아웃 버튼을 클릭 후 이동한 페이지에서 다시 message 라는 클래스 속성값을 가지고 있는 요소의 텍스트 정보를 출력

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time
import pyautogui

from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service = service, options = options)

url = "https://divjason.github.io/sellenium-test/"
driver.get(url)

element = driver.find_element(By.NAME, "username")
element.clear()
element.send_keys("error@error.com")

element = driver.find_element(By.NAME, "password")
element.clear()
element.send_keys("")

# element.send_keys(Keys.RETURN)
element_login = driver.find_element(By.CSS_SELECTOR, "input[type='submit']")

element_login.click()

time.sleep(3)

element_logout = driver.find_element(By.CSS_SELECTOR, "input[type='submit']")

element_logout.click()

time.sleep(3)

elements = driver.find_elements(By.CSS_SELECTOR, "div.message")

for element in elements :
    print(element.text)

# driver.save_screenshot("shot1.png")

driver.quit()
```
