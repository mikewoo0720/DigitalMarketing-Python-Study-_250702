```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

base_url = "https://davelee-fun.github.io/"
page_url_template = "https://davelee-fun.github.io/page{}"

# 1페이지는 base_url, 2페이지부터는 page_url_template 적용
for page_num in range(1, 9):
    if page_num == 1:
        url = base_url
    else:
        url = page_url_template.format(page_num)
    
    driver.get(url)
    time.sleep(1)  # 로딩 대기
    
    elements = driver.find_elements(By.CSS_SELECTOR, "h4.card-text")
    print(f"===== Page {page_num} =====")
    for element in elements:
        print(element.text.split(":")[1])

driver.quit()
```

```python
#특정 속성값으로 찾아오기
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service = service, options = options)

url = "http://davelee-fun.github.io/"
driver.get(url)

# element = driver.find_element(By.CSS_SELECTOR, "a.text-dark")
# print("text : ", element.text)
# print("deiver.title : ", element.get_attribute("href"))#href 속성값 찾아오기 ex) href 속성값을 가지고 다른 페이지 이동 다른페이지 크롤링 가능
# print("get_attribute('class') : ", element.get_attribute("class"))#class의 속성값 찾아오기
# print("get_attribute('text') : ", element.get_attribute("text"))#문자열은 항상 text 속성값을 가지고 있다는 의미 이다.


elements = driver.find_elements(By.CSS_SELECTOR, "a.text-dark")
for element in elements :
    print("text : ", element.text)
    print("deiver.title : ", element.get_attribute("href"))#href 속성값 찾아오기 ex) href 속성값을 가지고 다른 페이지 이동 다른페이지 크롤링 가능
    print("get_attribute('class') : ", element.get_attribute("class"))#class의 속성값 찾아오기
    print("get_attribute('text') : ", element.get_attribute("text"))#문자열은 항상 text 속성값을 가지고 있다는 의미 이다.



time.sleep(1)

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service = service, options = options)

url = "http://davelee-fun.github.io/"
driver.get(url)

# driver.excute_script("window.scrollTo(0, document.body.scrollHeight)")
height = driver.execute_script("return document.body.scrollHeight")
print("scrollHeight : ", height)

client_height = driver.execute_script("return document.documentElement.clientHeight")
print("clientHeight : ", client_height)

offset_height = driver.execute_script("return document.body.offsetHeight")
print("offsetHeight : ", offset_height)

time.sleep(1)

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time

from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service = service, options = options)

url = "http://davelee-fun.github.io/"
driver.get(url)

driver.execute_script("window.scrollTo(0, document.body.scrollHeight)")

element = driver.find_element(By.NAME, "EMAIL")#NAME 에서 EMAIL이라는 속성값을 찾아오자
element.clear()#찾은 인풋태그 안의 내용을 비운다(key 값을 입력하려면 비워줘야함)

element.send_keys("error@error.com")#이메일 입력

time.sleep(3)

element.send_keys(Keys.RETURN)

time.sleep(3)

driver.quit()
```

```python
#브라우저 화면 캡쳐
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.keys import Keys
import time
import pyautogui#우리가 사용하는 브라우저의 width 와 height 값을 가지고 있음, 구조분해할당 쓸수있음.

from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup

screen_width, screen_height = pyautogui.size()#구조분해할당으로 변수값 선언
print(screen_width, screen_height)

service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")
options = Options()
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")

driver = webdriver.Chrome(service = service, options = options)

url = "http://davelee-fun.github.io/"
driver.get(url)

driver.set_window_size(screen_width, screen_height)#선언한 사이즈 만큼의 윈도우 브라우저를 인식
element = driver.find_element(By.TAG_NAME, "body")

element.screenshot("samplesite.png")#브라우저 화면 요소를 캡쳐

time.sleep(3)

driver.quit()
```
