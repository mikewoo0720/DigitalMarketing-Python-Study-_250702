```python
#파라미터 값의 페이지 변화를 가지고 for문으로 찾아오기
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time

service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

base_url = "https://davelee-fun.github.io/"
page_url_template = "https://davelee-fun.github.io/page{}"

# 1페이지 base_url, 2페이지부터 page_url_template
for page_num in range(1, 9):
    if page_num == 1:
        url = base_url
    else:
        url = page_url_template.format(page_num)
    
    driver.get(url)
    time.sleep(1)
    
    elements = driver.find_elements(By.CSS_SELECTOR, "h4.card-text")
    print(f"===== Page {page_num} =====")
    for element in elements:
        print(element.text.split(":")[1])

driver.quit()
```

```python
#여러 페이지 데이터 크롤링(수정)
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
import time

# 크롬 드라이버 설정
service = Service(r"C:\Users\user\Downloads\디지털 마케팅\chromedriver\chromedriver.exe")

prefs = {
    "credentials_enable_service": False,
    "profile.password_manager_enabled": False,
    "profile.default_content_setting_values.notifications": 2
}

options = Options()
options.add_experimental_option("prefs", prefs)
options.add_argument("--disable-gpu")
options.add_argument("--start-maximized")
options.add_argument("--disable-save-password-bubble")
options.add_argument("--disable-infobars")
options.add_argument("--disable-extensions")

driver = webdriver.Chrome(service=service, options=options)

url = "https://davelee-fun.github.io/"
driver.get(url)
time.sleep(1)

# 8페이지 반복
for n in range(1, 9):
    print(f"\n--- Page {n} ---")
    
    # 페이지 내용 수집
    elements = driver.find_elements(By.CSS_SELECTOR, "h4.card-text")
    for element in elements:
        print(element.text)

    # 다음 페이지 클릭 (마지막 페이지는 건너뜀)
    if n < 8:
        element_nextpg = driver.find_element(By.CSS_SELECTOR, "div > a:nth-child(10)")
        driver.execute_script("arguments[0].scrollIntoView();", element_nextpg)
        time.sleep(1)
        element_nextpg.click()
        time.sleep(2)

driver.quit()
```
