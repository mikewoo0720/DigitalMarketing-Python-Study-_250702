```python
from selenium import webdriver #브라우저를 컨트롤 하기위한 로컬컴퓨터와 셀레니움의 드라이버가 일치하도록 조정
from selenium.webdriver.chrome.service import Service #데스크탑 기준, 가상 브라우저 버전, 윈도우 창 크기, 임시메모리 할당 등
from selenium.webdriver.chrome.options import Options #데스크탑 기준, 가상 브라우저 버전, 윈도우 창 크기, 임시메모리 할당 등
from selenium.webdriver.common.by import By #어떤 방식으로 찾아올지 설정 (실무에서 보통 id 값으로 찾아옴 없을경우 class나 선택자로 정 안되면 Xpath 마지막 tag (근데 가장 위험하다))
from webdriver_manager.chrome import ChromeDriverManager
import time #셀레니움은  동적사이트를 크롤링 하기위해 실제 동작하는 형식이라 구현되는 시간 조정을 위해
import re #링크 찾아오기 힘들어서 정규표현식 사용위해
import pandas as pd #엑셀저장하기 위해

service = Service(ChromeDriverManager().install())#브라우저 버전 일치화
options = Options()#옵션 설정을 위한 프로토타입 객체 생성
# options.add_argument("--headless")#인스턴스 객체로 옵션들 설정
# options.add_argument("--disable-gpu")#gpu 사용하지마라
options.add_argument("--disable-dev-shm-usage")#메모리 사용하지 마라
options.add_argument("--start-maximized")#시작시 브라우저 크기 최대
options.add_argument("--window-size=1920x1080")#윈도우 사이즈 알려주기
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/139.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")#우리는 한국어

driver = webdriver.Chrome(service=service, options=options)#가상 웹 브라우저를 띄우기 위한 프로토타입 객체(객체는 서비스와 옵션 값을 받아옴) 생성

category_urls = {#원하는 카테고리 풀 웹주소 딕셔너리 형태로 선언
    "여성의류":"https://display.wconcept.co.kr/rn/best?displayCategoryType=10101&gnbType=Y",
    "가방":"https://display.wconcept.co.kr/rn/best?displayCategoryType=10102&displaySubCategoryType=ALL&gnbType=Y",
    "신발":"https://display.wconcept.co.kr/rn/best?displayCategoryType=10103&displaySubCategoryType=ALL&gnbType=Y",
    "액세서리":"https://display.wconcept.co.kr/rn/best?displayCategoryType=10104&displaySubCategoryType=ALL&gnbType=Y"
}

all_data = []#모든 데이터 받을 준비

def extract_product_id_from_imgsrc(src: str) -> str | None :#함수 실행을 파라미터 내 특정 값을 가져와서 문자열이거나 없거나로 반환
    if not src:
        return None
    m = re.search(r"/(\d+)(?:[_\.][^/?]+)?(?:\?.*)?$", src)#레귤러 익스프레션 객체안에 로우스트링을 이용해 하나이상의 정숫값을 찾아와라
    return m.group(1) if m else None

def build_product_url(pid: str | None) -> str :
    return f"https://www.concept.co.kr/Product/{pid}" if pid else ""

for cat_name, url in category_urls.items() :
    driver.get(url)
    time.sleep(3)

    for _ in range(10) :
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)

    products = driver.find_elements(By.CSS_SELECTOR, "div.product-item.item.type-all")[:100]


    for product in products :
        try :
            name = product.find_element(By.CSS_SELECTOR, "span.prdc-title span.text.detail").text.strip()
            try :
                img = product.find_element(By.CSS_SELECTOR, "button.area-img img")
                src = img.get_attribute("src")
                pid = extract_product_id_from_imgsrc(src)
                url = build_product_url(pid)
                
            except :
                url = ""
            try :
                review_info = product.find_element(By.CSS_SELECTOR, "span.area-info span.text.cnt").text
                review_count = int(re.sub(r"[^\d]", "", review_info)) if review_info else 0
            except :
                review_count = 0
            try :
                rating_text = product.find_element(By.CSS_SELECTOR, "span.area-info em.score").text
                rating_avg = float(re.findall(r"[0-9.]+", rating_text)[0])
            except :
                rating_avg = 0.0

            all_data.append({
                "카테고리": cat_name,
                "상품명": name,
                "URL": url,
                "리뷰수": review_count,
                "평균평점": rating_avg,
            })
            
        except Exception as e :
            print("에러발생 :",e)
            continue

driver.quit()


df = pd.DataFrame(all_data)

df_top100 = df.sort_values(by="리뷰수", ascending=False).head(100).reset_index(drop=True)

category_summary = df.groupby("카테고리").agg({
    "리뷰수": "sum",
    "평균평점": "mean"
}).reset_index()

print(f"""
상위 100개 상품 :
{df_top100.head()}

카테고리별 통계 :
{category_summary}
""")
```
