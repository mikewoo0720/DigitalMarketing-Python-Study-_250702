```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from deep_translator import GoogleTranslator
import pandas as pd, time, re, random

# =========================
# ⚙️ 설정
# =========================
RANKING_URL = (
    "https://www.musinsa.com/main/musinsa/ranking?"
    "gf=A&storeCode=musinsa&sectionId=199&contentsId=&"
    "categoryCode=002000&ageBand=AGE_BAND_ALL&subPan=product&period=MONTHLY"
)
TARGET_URLS = 1000

SCROLL_STEPS_PER_BLOCK = 16
STEP_SLEEP = 0.6
BLOCKS_MAX = 200
WAIT_GROWTH_BASE = 6.0
WAIT_GROWTH_MAX = 20.0
NO_GAIN_BLOCK_LIMIT = 10

# =========================
# 🌍 번역기 설정
# =========================
translator = GoogleTranslator(source='ko', target='en')

def translate_auto(text):
    """한국어 → 영어 자동 번역 (deep-translator 사용 + 예외 처리)"""
    if not text:
        return None
    try:
        result = translator.translate(text)
        return result
    except Exception:
        t = text.strip()
        if any(x in t for x in ["남", "남성", "남자"]):
            return "Male"
        elif any(x in t for x in ["여", "여성", "여자"]):
            return "Female"
        elif any(x in t for x in ["공용", "유니섹스"]):
            return "Unisex"
        return text

# =========================
# 🧩 숫자 변환 함수
# =========================
def normalize_korean_count(text):
    if not text or not isinstance(text, str):
        return None
    t = text.replace(",", "").strip()
    m = re.search(r"([\d]+(?:\.\d+)?)", t)
    if not m:
        return None
    num = float(m.group(1))
    if "만" in t:
        num *= 10000
    elif "천" in t:
        num *= 1000
    return int(num)

def safe_get_text(driver, selector):
    try:
        return driver.find_element(By.CSS_SELECTOR, selector).text.strip()
    except:
        return None

def get_spec_value(driver, label):
    try:
        dl_elems = driver.find_elements(By.CSS_SELECTOR, "dl > div")
        for div in dl_elems:
            try:
                dt = div.find_element(By.CSS_SELECTOR, "dt").text.strip()
                if label in dt:
                    return div.find_element(By.CSS_SELECTOR, "dd").text.strip()
            except:
                continue
    except:
        return None
    return None

# =========================
# 🚀 드라이버 설정
# =========================
options = Options()
options.add_argument("--window-size=1500,1000")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--no-sandbox")
# options.add_argument("--headless")
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

# =========================
# 🧭 URL 수집 (무한 스크롤)
# =========================
driver.get(RANKING_URL)
time.sleep(8)

collected = set()
no_gain_blocks = 0
growth_wait = WAIT_GROWTH_BASE

def current_links():
    hrefs = driver.execute_script("""
        return Array.from(document.querySelectorAll('a[href*="/products/"]'))
                     .map(a => a.href);
    """)
    return [h.split("?")[0] for h in hrefs if "/products/" in h]

print("📜 랭킹 페이지에서 URL 수집 시작")

for block in range(1, BLOCKS_MAX + 1):
    for step in range(SCROLL_STEPS_PER_BLOCK):
        scroll_value = random.uniform(0.6, 1.0)
        driver.execute_script(f"window.scrollBy(0, window.innerHeight * {scroll_value});")
        time.sleep(STEP_SLEEP)
        for u in current_links():
            collected.add(u)

    base_count = len(collected)
    waited = 0.0
    while waited < growth_wait:
        time.sleep(1.2)
        waited += 1.2
        for u in current_links():
            collected.add(u)
        if len(collected) > base_count:
            break

    gain = len(collected) - base_count
    print(f"📦 블록 {block}/{BLOCKS_MAX} ▶ 누적 {len(collected)}개 (+{gain})")

    if gain == 0:
        no_gain_blocks += 1
        growth_wait = min(WAIT_GROWTH_MAX, growth_wait + 3.0)
    else:
        no_gain_blocks = 0
        growth_wait = WAIT_GROWTH_BASE

    if len(collected) >= TARGET_URLS:
        print("✅ 목표 달성 (1000개 수집 완료)")
        break
    if no_gain_blocks >= NO_GAIN_BLOCK_LIMIT:
        print("⚠️ 증가 없음 → 스크롤 중단")
        break

product_urls = list(collected)[:TARGET_URLS]
print(f"🎯 URL 수집 완료: {len(product_urls)}개")

# =========================
# 💬 상세 정보 수집
# =========================
data = []

for idx, url in enumerate(product_urls, start=1):
    for attempt in range(3):
        try:
            driver.get(url)
            time.sleep(3.0)
            driver.execute_script("window.scrollBy(0, window.innerHeight * 0.5);")
            time.sleep(1.5)

            gender_ko = get_spec_value(driver, "성별")
            gender_en = translate_auto(gender_ko)

            row = {
                "product_name": safe_get_text(driver, "#root div.GoodsName__Wrap-sc-1omefes-0 span"),
                "url": url,
                "product_id": get_spec_value(driver, "품번"),
                "gender_ko": gender_ko,
                "gender_en": gender_en,
                "season": get_spec_value(driver, "시즌"),
                "monthly_views": normalize_korean_count(get_spec_value(driver, "조회수")),
                "sales": normalize_korean_count(get_spec_value(driver, "누적판매")),
            }

            data.append(row)
            print(f"✅ {idx}/{len(product_urls)}")
            break
        except Exception as e:
            print(f"⚠️ {idx}번째 실패 ({attempt+1}/3): {e}")
            time.sleep(4)
            if attempt == 2:
                print("🚫 최대 재시도 초과 — 건너뜀")

    if idx % 50 == 0:
        pd.DataFrame(data).to_excel("musinsa_partial_save_outer.xlsx", index=False)
        print(f"💾 {idx}개 중간 저장 완료")

# =========================
# 5️⃣ 저장 및 종료
# =========================
driver.quit()
pd.DataFrame(data).to_excel("C02_1000_outer.xlsx", index=False)
print("\n💾 C02_1000_outer.xlsx 완료 ✅")
```
