#### 문자열 관련 함수 리스트
- count : 문자열에 있는 특정 문자 갯수 세기
- index : 문자열에 있는 특정 문자의 인덱스 위치를 알려주기
- find : 문자열에 있는 특정 문자의 인덱스를 찾아오긴 하나, 만약 존재하지 않는다면 에러를 발생하지 않고 -1을 발생!
- join : 문자열 사이에 다른 문자 끼워넣기
- strip() : 문자열 앞 뒤에 공백을 지워주기(앞에서 부터 숫자, 횟수에 상관없이 끝까지)
- upper : 소문자를 대문자로 변경하기
- lower : 대문자를 소문자로 변경하기
- split : 문자열 내 특정 문자를 기준으로 나누기
- replace : 

```python
data = "Dave David"
data.count("Dav")
```

```python
string = "Dave ID is dave"

first_index = string.index("e")
string.index("e", first_index + 1)
```

```python
string01 = "Dave ID is dave"
string01.find("v")
string01.find("a")

if string01.find("x") == -1 :
    print("x는 string01 문자열 객체 안에 없습니다.")
```

```python
string02 = "12345"
comma = ","
comma.join(string02)
```

```python
data01 = " Dave "
data01.strip()

data02 = "      Dave02 "
data02.strip()

data03 = "     Dave03 "
data03.lstrip()

data04 = "     Dave04 "
data04.rstrip()
```

```python
string04 = "Dave ID is dave"
string04.upper()
```

```python
string05 = "Dave ID is dave"
string05.lower()
```

```python
string06 = "Dave goes to Korea"
string06.split()[3]
string06.split(" ")

string07 = "Dave/goes/to/Korea"
string07.split("/")
```
