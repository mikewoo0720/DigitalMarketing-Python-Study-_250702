```python
import time
import re
import pandas as pd
import chromedriver_autoinstaller

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import (
    TimeoutException,
    NoAlertPresentException,
    UnexpectedAlertPresentException,
)
from tqdm import tqdm


# ============================
# 0. 경로 설정
# ============================

EXCEL_PATH = "babitalk_event_urls_part1.xlsx"   # event_url 리스트 엑셀
OUTPUT_CSV = "babitalk_detail_full.csv"         # 결과 저장 파일명

MAX_ROWS = None        # None이면 전체, 숫자 넣으면 상위 N개만 테스트용
SAVE_INTERVAL = 50     # 중간 저장 주기


# ============================
# 1. CSS 선택자 (안정화 버전)
# ============================

# 시술명
TREATMENT_NAME_SELECTOR = "h1.text-label-common_5"

# 병원명
HOSPITAL_NAME_SELECTOR = "p.text-label-common_4.cursor-pointer"

# 평점 / 후기
RATING_SELECTOR = "div.flex.items-center.gap-\\[10px\\] h5"
REVIEW_COUNT_SELECTOR = "div.flex.items-center.gap-\\[10px\\] button"

# 가격 정보 (정가)
ORIGINAL_PRICE_SELECTORS = [
    "h6.text-label-skinTreatment_2",  # 피부/레이저 시술 정가
    "h6.text-label-plasticsurgery_2", # 성형 시술 정가
    "h2 span",                        # 할인 없는 경우 가격
]

# 할인가
SALE_PRICE_SELECTOR = "h2.flex.items-center.gap-\\[2px\\]"

# 할인율: 피부 / 성형 모두 고려
DISCOUNT_RATE_SELECTORS = [
    "h2.text-label-skinTreatment_2",
    "h2.text-label-plasticsurgery_2",
]

# 이벤트 설명
EVENT_DESC_SELECTOR = "#detail div.bg-white div div"

# 이벤트 정보 박스
EVENT_PERIOD_SELECTOR = "#detail .bg-neutral-100 div:nth-child(1) .text-label-common_3"
EVENT_TARGET_SELECTOR = "#detail .bg-neutral-100 div:nth-child(2) .text-label-common_3"
SIDE_EFFECT_SELECTOR  = "#detail .bg-neutral-100 div:nth-child(3) .text-label-common_3"

# 병원 주소
HOSPITAL_ADDRESS_SELECTOR = (
    "#hospital div.flex.flex-col.justify-start.items-start.gap-\\[4px\\] h5:nth-child(3)"
)

# "시술 정보 더보기" 버튼
MORE_DETAIL_BTN_SELECTOR = "#btn_more_detail"


# ============================
# 2. 공용 함수
# ============================

def init_driver():
    """크롬 드라이버 초기화"""
    chromedriver_autoinstaller.install()
    options = Options()
    options.add_argument("--start-maximized")
    # options.add_argument("--headless")  # 서버에서 돌릴 때는 주석 해제
    driver = webdriver.Chrome(options=options)
    return driver


def safe_text(driver, selector):
    """CSS 셀렉터로 요소 찾고 text만 안전하게 리턴 (없으면 None)"""
    try:
        el = driver.find_element(By.CSS_SELECTOR, selector)
        return el.text.strip()
    except Exception:
        return None


def first_non_empty(driver, selectors):
    """여러 셀렉터 중에서 먼저 잡히는 텍스트 하나 리턴"""
    for s in selectors:
        txt = safe_text(driver, s)
        if txt:
            return txt
    return None


def extract_number(text):
    """문자열에서 숫자만 추출해서 int로 리턴"""
    if not text:
        return None
    num = re.sub(r"[^\d]", "", text)
    return int(num) if num else None


def handle_alert_if_exists(driver):
    """알럿 있으면 닫고 True, 없으면 False"""
    try:
        alert = driver.switch_to.alert
        print(f"[알럿] {alert.text}")
        alert.accept()
        return True
    except Exception:
        return False


def click_more_detail_if_exists(driver):
    """'시술 정보 더보기' 버튼이 있으면 JS로 클릭"""
    try:
        btns = driver.find_elements(By.CSS_SELECTOR, MORE_DETAIL_BTN_SELECTOR)
        if not btns:
            return False

        btn = btns[0]
        driver.execute_script("arguments[0].scrollIntoView({block:'center'});", btn)
        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.5)

        handle_alert_if_exists(driver)
        return True
    except UnexpectedAlertPresentException:
        handle_alert_if_exists(driver)
        return False
    except Exception:
        return False


def parse_discount_rate(driver):
    """
    할인율 파싱:
    1) skinTreatment_2 / plasticsurgery_2 클래스 우선
    2) 그래도 못 찾으면 % 포함된 h2를 fallback으로 검색
    """
    # 1차: 명시적인 클래스들
    for sel in DISCOUNT_RATE_SELECTORS:
        txt = safe_text(driver, sel)
        if txt and "%" in txt:
            return txt.strip()

    # 2차: 페이지 내 모든 h2 중 '%' 포함된 것 찾기
    try:
        h2s = driver.find_elements(By.CSS_SELECTOR, "h2")
        for h in h2s:
            t = h.text.strip()
            if "%" in t:
                return t
    except Exception:
        pass

    return None


# ============================
# 3. 단일 페이지 크롤링
# ============================

def crawl_treatment_page(driver, url: str) -> dict:
    driver.get(url)

    # 시술명이 실제로 렌더링될 때까지 기다리기
    try:
        WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.CSS_SELECTOR, TREATMENT_NAME_SELECTOR))
        )
    except TimeoutException:
        print(f"❗ [Timeout] 시술명 안 뜸 → {url}")
        return {"treatment_url": url}

    time.sleep(0.3)

    # 시술명 / 병원명 / 주소
    treatment_name   = safe_text(driver, TREATMENT_NAME_SELECTOR)
    hospital_name    = safe_text(driver, HOSPITAL_NAME_SELECTOR)
    hospital_address = safe_text(driver, HOSPITAL_ADDRESS_SELECTOR)

    # 평점 / 후기
    rating_text = safe_text(driver, RATING_SELECTOR)
    try:
        rating = float(rating_text) if rating_text else None
    except ValueError:
        rating = None

    review_raw = safe_text(driver, REVIEW_COUNT_SELECTOR)
    review_num = extract_number(review_raw)
    review_count = f"{review_num}개" if review_num is not None else None

    # 가격 정보
    original_price = first_non_empty(driver, ORIGINAL_PRICE_SELECTORS)
    discount_rate  = parse_discount_rate(driver)
    sale_price     = safe_text(driver, SALE_PRICE_SELECTOR)

    # 이벤트 / 부작용 정보
    click_more_detail_if_exists(driver)
    time.sleep(0.3)

    event_desc   = safe_text(driver, EVENT_DESC_SELECTOR)
    event_period = safe_text(driver, EVENT_PERIOD_SELECTOR)
    event_target = safe_text(driver, EVENT_TARGET_SELECTOR)
    side_effect  = safe_text(driver, SIDE_EFFECT_SELECTOR)

    row = {
        "treatment_url": url,
        "treatment_name": treatment_name,
        "rating": rating,
        "review_count": review_count,
        "original_price_text": original_price,
        "discount_rate_text": discount_rate,
        "sale_price_text": sale_price,
        "event_description": event_desc,
        "event_period": event_period,
        "event_target": event_target,
        "side_effect_info": side_effect,
        "hospital_name": hospital_name,
        "hospital_address": hospital_address,
    }

    # 모든 값이 None이면 로그
    if all(v is None for k, v in row.items() if k != "treatment_url"):
        print(f"⚠️ [모든 필드 None] {url}")

    return row


# ============================
# 4. 메인 – 전체 URL 크롤링
# ============================

def main():
    # 1) event_url 엑셀 불러오기
    df = pd.read_excel(EXCEL_PATH)

    if "event_url" not in df.columns:
        raise ValueError("엑셀에 'event_url' 컬럼이 없습니다. 컬럼명을 확인해주세요.")

    urls = df["event_url"].dropna().astype(str).tolist()
    if MAX_ROWS is not None:
        urls = urls[:MAX_ROWS]

    print(f"크롤링 대상 URL 수: {len(urls)}")

    # 2) 드라이버 띄우기
    driver = init_driver()
    results = []

    try:
        for i, url in enumerate(tqdm(urls, desc="바비톡 시술 크롤링", unit="건"), start=1):
            try:
                row = crawl_treatment_page(driver, url)
                results.append(row)
            except Exception as e:
                print(f"❗ [에러] idx={i}, url={url} / {e}")

            # 중간 저장
            if i % SAVE_INTERVAL == 0:
                pd.DataFrame(results).to_csv(
                    OUTPUT_CSV, index=False, encoding="utf-8-sig"
                )

            time.sleep(1)   # 서버 부하 방지용
    finally:
        driver.quit()

    # 최종 저장
    pd.DataFrame(results).to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")
    print(f"\n✅ 크롤링 완료 – 총 {len(results)}건 저장 → {OUTPUT_CSV}")


if __name__ == "__main__":
    main()
```
