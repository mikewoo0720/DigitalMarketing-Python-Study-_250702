```python
import time
import re

import pandas as pd
import chromedriver_autoinstaller
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import (
    TimeoutException,
    NoAlertPresentException,
    UnexpectedAlertPresentException,
)
from tqdm import tqdm  # 진행률 바


# =========================
# 0. 경로 & 기본 설정
# =========================

########################OUTPUT_CSV 이름 뒤에 ex) _sy처럼 수정 부탁드려욥

EXCEL_PATH = "babitalk_event_urls.xlsx"  # 🔧 필요 시 수정
OUTPUT_CSV = "babitok_detail_procedure.csv"

MAX_ROWS = None      # None = 전체, 숫자 넣으면 상위 N개만
SAVE_INTERVAL = 50   # 50개마다 중간 저장


# =========================
# 1. 셀렉터 모음
# =========================

# 정가 셀렉터 (할인 O / 할인 X 둘 다 커버)
ORIGINAL_PRICE_SELECTORS = [
    # 할인 있는 시술의 정가
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(3) > div:nth-child(1) > h6",
    # 할인 없는 시술의 정가
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(3) > div > div.flex.items-center.gap-\\[6px\\] > h2 > span",
]

# 할인율 (텍스트)
DISCOUNT_RATE_SELECTOR = (
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(3) > div:nth-child(2) > div.flex.items-center.gap-\\[6px\\] > "
    "h2.text-label-skinTreatment_2.leading-\\[150\\%\\].text-inherit.text-xl.font-semibold"
)

# 할인가 (텍스트)
SALE_PRICE_SELECTOR = (
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(3) > div:nth-child(2) > div.flex.items-center.gap-\\[6px\\] > "
    "h2.flex.items-center.gap-\\[2px\\].leading-\\[150\\%\\].text-inherit.text-xl.font-semibold"
)

# 시술명
TREATMENT_NAME_SELECTOR = (
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(1) > div.flex.flex-col > h1"
)

# 평점
RATING_SELECTOR = (
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(1) > div.flex.items-center.gap-\\[10px\\] > div > h5"
)

# 후기 갯수
REVIEW_COUNT_SELECTOR = (
    "#content-body > div > div.flex.flex-col.gap-\\[16px\\].w-\\[100\\%\\].pb-\\[18px\\] > "
    "div:nth-child(1) > div.flex.items-center.gap-\\[10px\\] > button"
)

# 이벤트 설명
EVENT_DESC_SELECTOR = (
    "#detail > div.w-full.bg-white.flex-col.justify-start.items-start.inline-flex.mb-\\[16px\\] > div > div"
)

# "시술 정보 더보기" 버튼
MORE_DETAIL_BTN_SELECTOR = "#btn_more_detail"

# 이벤트 정보 박스 베이스
EVENT_INFO_BASE = "#detail div.bg-neutral-100"

EVENT_PERIOD_SELECTOR = EVENT_INFO_BASE + " > div:nth-child(1) .text-label-common_3"
EVENT_TARGET_SELECTOR = EVENT_INFO_BASE + " > div:nth-child(2) .text-label-common_3"
SIDE_EFFECT_SELECTOR = EVENT_INFO_BASE + " > div:nth-child(3) .text-label-common_3"

# 병원명 div (arrow_forward_ios까지 포함된 텍스트)
HOSPITAL_NAME_SELECTOR = (
    "#hospital > div.cursor-pointer > div > div.flex.flex-col.justify-start.items-start.gap-\\[4px\\].w-full "
    "> div.flex.justify-start.items-center.gap-\\[8px\\].cursor-pointer"
)

# ✅ 병원 주소 셀렉터 (사용자가 준 것)
HOSPITAL_ADDRESS_SELECTOR = (
    "#hospital > div.cursor-pointer > div > div.flex.flex-col.justify-start.items-start.gap-\\[4px\\].w-full "
    "> div:nth-child(2) > h5:nth-child(3)"
)


# =========================
# 2. 유틸 함수
# =========================

def init_driver():
    chromedriver_autoinstaller.install()
    options = Options()
    options.add_argument("--start-maximized")
    # options.add_argument("--headless")  # 필요하면 주석 해제
    driver = webdriver.Chrome(options=options)
    return driver


def safe_text(driver, selector):
    try:
        el = driver.find_element(By.CSS_SELECTOR, selector)
        return el.text.strip()
    except Exception:
        return None


def first_non_empty(driver, selectors):
    for sel in selectors:
        if not sel:
            continue
        txt = safe_text(driver, sel)
        if txt:
            return txt
    return None


def extract_number(text):
    if not text:
        return None
    digits = re.sub(r"[^\d]", "", text)
    return int(digits) if digits else None


def normalize_price_text(text):
    """가격 텍스트 뒤에 '원'이 없으면 붙이기"""
    if not text:
        return None
    t = text.strip()
    if not t.replace(" ", "").replace("\n", "").endswith("원"):
        t = t + "원"
    return t


def normalize_discount_text(text):
    """할인율 텍스트 뒤에 '%'가 없으면 붙이기"""
    if not text:
        return None
    t = text.strip()
    if not t.endswith("%"):
        t = t + "%"
    return t


def normalize_hospital_name(text):
    """병원명에서 arrow_forward_ios 제거"""
    if not text:
        return None
    t = text.replace("arrow_forward_ios", "").strip()
    return t


def handle_alert_if_exists(driver):
    """알럿(이벤트 정보가 없습니다 등)이 뜨면 닫고 True 리턴, 없으면 False"""
    try:
        alert = driver.switch_to.alert
        text = alert.text
        print(f"[알럿 감지] {text}")
        alert.accept()
        return True
    except NoAlertPresentException:
        return False


def click_more_detail_if_exists(driver):
    """
    '시술 정보 더보기' 버튼 클릭.
    - JS로 강제 클릭해서 click intercepted 회피
    - 알럿(이벤트 정보 없음) 뜨면 닫고 False 리턴
    - 실패해도 예외 안 던지고 False 리턴
    """
    try:
        buttons = driver.find_elements(By.CSS_SELECTOR, MORE_DETAIL_BTN_SELECTOR)
        if not buttons:
            return False

        btn = buttons[0]
        driver.execute_script("arguments[0].scrollIntoView({block: 'center'});", btn)
        time.sleep(0.3)

        driver.execute_script("arguments[0].click();", btn)
        time.sleep(0.5)

        if handle_alert_if_exists(driver):
            return False

        return True

    except UnexpectedAlertPresentException:
        handle_alert_if_exists(driver)
        return False
    except Exception as e:
        print(f"[더보기 클릭 실패] {e}")
        handle_alert_if_exists(driver)
        return False


def parse_price_block_text_only(driver):
    """정가 / 할인율 / 할인가 텍스트만 추출 + 포맷 정리"""
    original_price_text = first_non_empty(driver, ORIGINAL_PRICE_SELECTORS)
    discount_rate_text = safe_text(driver, DISCOUNT_RATE_SELECTOR)
    sale_price_text = safe_text(driver, SALE_PRICE_SELECTOR)

    original_price_text = normalize_price_text(original_price_text)
    sale_price_text = normalize_price_text(sale_price_text)
    discount_rate_text = normalize_discount_text(discount_rate_text)

    return {
        "original_price_text": original_price_text,
        "discount_rate_text": discount_rate_text,
        "sale_price_text": sale_price_text,
    }


# =========================
# 3. 단일 페이지 크롤링
# =========================

def crawl_treatment_page(driver, url):
    driver.get(url)

    # 메인 콘텐츠 로딩 대기 + 약간의 여유
    WebDriverWait(driver, 12).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "#content-body"))
    )
    time.sleep(1)

    # 시술명
    treatment_name = safe_text(driver, TREATMENT_NAME_SELECTOR)

    # 평점
    rating_text = safe_text(driver, RATING_SELECTOR)
    try:
        rating = float(rating_text) if rating_text else None
    except ValueError:
        rating = None

    # 후기 갯수 → 숫자 추출 후 "개" 붙이기
    review_count_raw = safe_text(driver, REVIEW_COUNT_SELECTOR)
    review_count_num = extract_number(review_count_raw)
    review_count = f"{review_count_num}개" if review_count_num is not None else None

    # 가격 정보
    price_info = parse_price_block_text_only(driver)

    # 시술 정보 더보기 (이벤트 영역 노출)
    click_more_detail_if_exists(driver)
    time.sleep(0.8)

    # 이벤트 / 병원 정보
    event_desc = safe_text(driver, EVENT_DESC_SELECTOR)
    event_period = safe_text(driver, EVENT_PERIOD_SELECTOR)
    event_target = safe_text(driver, EVENT_TARGET_SELECTOR)
    side_effect = safe_text(driver, SIDE_EFFECT_SELECTOR)

    hospital_name_raw = safe_text(driver, HOSPITAL_NAME_SELECTOR)
    hospital_name = normalize_hospital_name(hospital_name_raw)

    # ✅ 병원 주소
    hospital_address = safe_text(driver, HOSPITAL_ADDRESS_SELECTOR)

    row = {
        "treatment_url": url,
        "treatment_name": treatment_name,
        "rating": rating,
        "review_count": review_count,  # 예: "29개"
        "original_price_text": price_info["original_price_text"],
        "discount_rate_text": price_info["discount_rate_text"],
        "sale_price_text": price_info["sale_price_text"],
        "event_description": event_desc,
        "event_period": event_period,
        "event_target": event_target,
        "side_effect_info": side_effect,
        "hospital_name": hospital_name,
        "hospital_address": hospital_address,
    }

    return row


# =========================
# 4. 메인 실행부
# =========================

def main():
    df = pd.read_excel(EXCEL_PATH)
    if "event_url" not in df.columns:
        raise ValueError("엑셀에 'event_url' 컬럼이 없습니다. 컬럼명을 확인해주세요.")

    if MAX_ROWS is None:
        urls = df["event_url"].tolist()
    else:
        urls = df["event_url"].head(MAX_ROWS).tolist()

    driver = init_driver()
    results = []

    try:
        for i, url in enumerate(tqdm(urls, desc="바비톡 시술 크롤링", unit="건"), start=1):
            try:
                row = crawl_treatment_page(driver, url)
                results.append(row)
            except Exception as e:
                print(f"\n⚠️ 에러 발생 (idx={i}, url={url}): {e}")
                continue

            # 중간 저장
            if i % SAVE_INTERVAL == 0 or i == len(urls):
                temp_df = pd.DataFrame(results)
                temp_df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")

            time.sleep(1.5)  # 사이트 부담 줄이기

    finally:
        driver.quit()

    print(f"\n✅ 크롤링 완료. 총 {len(results)}건 저장 → {OUTPUT_CSV}")


if __name__ == "__main__":
    main()
```
