```python
# numpy => 파이썬에서 수치계산 및 배열연산을 보다 효율적으로 하고자 할 때 사용가능한 라이브러리
# 객체지향 프로그래밍 언어 => [] : 배열 || [] : 리스트
# 1) 파이썬 내 자료구조 중 하나인 리스트보다 훨씬 빠르고, 메모리 관리 효과적인 연산 처리
# 2) 수학함수 제공 : 선형대수, 난수생성 등의 수학적 기능을 가져다가 사용할 때 넘파이 내부에 내장된 함수
# 3) 현재 파이썬에서 사용가능한 머신러닝 기초 모델들을 수학적으로 표현할 때, 넘파이
# 4) 시뮬레이션, 통계분석 : 넘파이 내부에 있는 함수들을 많이 사용

import plotly.graph_objects as go
import numpy as np

x_data = np.random.rand(100)
y_data = np.random.rand(100)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=x_data,
        y=y_data,
        mode="markers",
        # 딕셔너리 리터럴 방식
        # marker={
        #     "size": 8,
        #     "color": "#00F",
        #     "opacity": 0.7
        # },
        # dict() 수동 형변환 함수를 활용한 방식
        marker = dict(
            size = 8,
            color = "#00F",
            opacity = 0.7
        ),
        name="산점도 데이터",
        showlegend=True
    )
)

fig.update_layout(
    title="산점도 그래프 예제",
    x=
)

fig.show()
# x_list = [1, 2, 3]
# print(x_data)
# print(type(x_data))
# print(type(x_list))
```

```python
import plotly.graph_objects as go

categories = ["A", "B", "C", "D"]
values = [23, 17, 35, 29]

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=categories,
        y=values,
        name="막대 그래프 데이터",
        showlegend=True
    )
)

fig.update_layout(
    title="막대 그래프 예제",
    xaxis_title="범주",
    yaxis_title="값",
)
```

```python
import plotly.graph_objects as go
import numpy as np

data =  np.random.rand(500)

fig = go.Figure()

fig.add_trace(
    go.Histogram(
        x=data,
        # Number of bins in x-axis => bin 구간
        nbinsx=30,
        name="히스토그램 데이터",
        showlegned=True
    )
)

fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

data1 = np.random.rand(100)
data2 = np.random.rand(100)

fig = go.Figure()

fig.add_trace(
    go.Box(
        y=data1,
        name="비교할 열1"
    )
)

fig.add_trace(
    go.Box(
        y=data2,
        name="비교할 열2"
    )
)

fig.update_layout(
    title="상자 그림 예제",
    yaxis_title="값"
)

fig.show()
```

```python
# Plotly Graph Objects 3개의 요소를 활용해서 그래프의 옵션을 정의
# 1) update_layout()
# 2) update_xaxes()
# 3) update_yaxes()

# numpy => 파이썬에서 수치계산 및 배열연산을 보다 효율적으로 하고자 할 때 사용가능한 라이브러리
# 객체지향 프로그래밍 언어 => [] : 배열 || [] : 리스트
# 1) 파이썬 내 자료구조 중 하나인 리스트보다 훨씬 빠르고, 메모리 관리 효과적인 연산 처리
# 2) 수학함수 제공 : 선형대수, 난수생성 등의 수학적 기능을 가져다가 사용할 때 넘파이 내부에 내장된 함수
# 3) 현재 파이썬에서 사용가능한 머신러닝 기초 모델들을 수학적으로 표현할 때, 넘파이
# 4) 시뮬레이션, 통계분석 : 넘파이 내부에 있는 함수들을 많이 사용

import plotly.graph_objects as go
import numpy as np

x_data = np.random.rand(100)
y_data = np.random.rand(100)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=x_data,
        y=y_data,
        mode="markers",
        # 딕셔너리 리터럴 방식
        # marker={
        #     "size": 8,
        #     "color": "#00F",
        #     "opacity": 0.7
        # },
        # dict() 수동 형변환 함수를 활용한 방식
        marker = dict(
            size = 8,
            color = "#00F",
            opacity = 0.7
        ),
        name="산점도 데이터",
        showlegend=True
    )
)

# fig.update_layout(
#     title="산점도 그래프 예제",
#     xaxis_title="x축",
#     yaxis_title="y축",
# )

fig.update_layout(
    title={
        "text": "새로운 그래프 제목",
        "x": 0.5,
        "font": {
            "size":24,
            "color": "darkblue"
        }
    }
)
fig.show()
# x_list = [1, 2, 3]
# print(x_data)
# print(type(x_data))
# print(type(x_list))
```

```python
import plotly.graph_objects as go
import numpy as np

x_data = np.random.rand(100)
y_data = np.random.rand(100)
categories = np.random.choice(["A","B","C"], 100)

fig = go.Figure()

for cat in ["A","B","C"] :
    fig.add_trace(
        go.Scatter(
            x=x_data[categories==cat],
            y=y_data[categories==cat],
            mode="markers",
            name = f"범주{cat}"
        )
    )

    fig.update_layout(
        title="범례 위치 예제",
        legend={
            "title": "범례 제목",
            "x": 1.02,
            "y": 1,
            "xanchor": "left",
            "yanchor": "top",
            "bordercolor": "#000",
            "borderwidth": 1
        }
    )

fig.show()
```

```python
import plotly.graph_objects as go
import numpy as np

x_data = np.random.rand(50)
y_data = np.random.rand(50)

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=x_data,
        y=y_data,
        mode="lines",
        mode="lines",
        showlegend=True
    )
)

fig.update_layout(
    title="x축 설정 예제"
)

fig.update_xaxes(
    title_text="시간 (초)",
    # 세부적인 속성값을 직접 딕셔너리 형태 => 딕셔너리 리터럴
    # title_font={
    #     "size": 18,
    #     "color": "purple"
    # }
    title_font=dict(size=18, color="purple"),
    tickangle=45,
    dtick=5
    
)

fig.show()
```

```python
import plotly.graph_objects as go
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "A":np.random.rand(50)
})

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=df.index,
        y=df["A"],
        mode="lines+markers",
        name="A 데이터",
        showlegend=True
    )
)

fig.update_layout(
    title={
        "text": "Graph with go.Scatter",
        "font": {"size": 16}
    },
    # xaxis_title="Index"
    # yaxis_title="A"
    xaxis={"title": "Index"},
    yaxis={"title": "A"}
)

fig.show()
```
