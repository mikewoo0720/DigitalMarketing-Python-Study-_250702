7. 미분
- 그래프 내 한 점에서의 기울기를 미분
- 실제값 = 예상값 (딥러닝 학습.예측)
- y = 5x + 1 => y' = 5
- y = 5x^2 + 1 => y' = 10x
- 상미분 : x (독립변수) 1개
- 편미분 : x (독립변수) 2개 이상
- y = 5a^2 + 2b + 1 => 변수 2개

8. 딥러닝의 예측값, 향상 시키기 위한 방법
- 가중치 (기울기) + 편향 (절편) => 실제값 & 예측(상)값
- 실제값 - 예측값 = 잔차 (residual)
- 비용함수 (cost function) : 잔차의 값을 최소화할 수 있도록 해주는 함수

9. 비용함수에서 사용가능한 기법
- MAE : 평균절대오차
- *MSE : 평균제곱근오차
- RMSE : 평균제곱근편차
- RMSLE : 평균제곱근대수오차

10. 최소제곱법
- 결국 잔차를 구하는 공식 => 2차 함수 => 독립변수 2개 (가중치, 편향)
- 기울기 계산 => 미분 => 편미분 계산
- 실제값 = 예측값 => 편미분 = 0
- 실제값과 가장 유사한 w, b

11. 경사하강법 (Gradient Descent)
- 실제값 & 예측값의 잔차가 최소값이 될 수 있을 때까지 순차적으로 값을 낮춰가는 수학적 연산 기법 | 알고리즘

torch.zeros(x, y, z) : 0의 값으로 shape 모양의 텐서 생성
torch.ones(x, y, z) : 1의 값으로 shape 모양의 텐서 생성
torch.rand(x, y, z) : 0부터 1사이에 균일한 확률 분포로 랜덤값을 가진 (x, y, z) shape 모양의 텐서 생성
torch.randn(x, y, z) : 평균 0, 표준편차가 1이면서, 균일한 확률 분포로 랜덤값을 가진 (x, y, z) shape 모양의 텐서 생성
torch.full((x, y, z), a) : a값으로, shape 모양의 텐서 생성

torch.zeros_like(tensor) : 인자값으로 들어온 tensor 객체와 동일한 shape를 가지되, 각 원소값은 0으로 이루어진 텐서 생성
torch.ones_like(tensor) : 인자값으로 들어온 tensor 객체와 동일한 shape를 가지되, 각 원소값은 1로 이루어진 텐서 생성
torch.rand_like(tensor) : 인자값으로 들어온 tensor 객체와 동일한 shape를 가지되, 0부터 1사이에 균일한 확률 분포로 랜덤값을 가진 텐서 생성
torch.randn_like(tensor) : 인자값으로 들어온 tensor 객체와 동일한 shape를 가지되, 표준편차가 1이면서, 균일한 확률 분포로 랜덤값을 가진 텐서 생성
torch.full_like(tensor) : 인자값으로 들어온 tensor 객체와 동일한 shape를 가지되, 각 원소값이 a값으로 텐서 생성

torch.empty(x, y, z) : 초기화 되지 않은 값을 가진 (x, y) shape의 행렬를 생성

```python
# 텐서 연산 (브로드캐스팅)
# Java, C, TS : 업캐스팅, 다운캐스팅
# 상대적으로 큰 값을 처리할 수 있는 변수 > 작은 값을 처리할 수 있는 변수 할당
# 브로드 (Broad) : Broaden 동사의 파생명사
# 3 x 1 : 3행 1렬
# 1 x 3 : 1행 3렬
# 3 x 3 : 브로드캐스팅
# 서로 다른 원소의 개수를 가지고 있는 두개의 텐서가 존재하는데, 이 중 특정 차원의 원소 수가 1이거나, 존재하지 않는 경우에는 브로드캐스팅이 발생되었을 때, 차원이 확대가 될 수 있다
```
