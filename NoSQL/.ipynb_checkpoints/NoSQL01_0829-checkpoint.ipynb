- NoSQL 은 schema & JOIN 등의 문법 및 관리시스템이 존재하지 x

#### 5. MongoDB의 주요 데이터 타입
- String : 문자열 데이터 // name : "David"
- Integer : 정수형 데이터 // age : 20
- Boolean : 논리형 데이터 // isStudent : true
- Double : 실수자료형 데이터 (*부동소수점 형태의 값) // rating : 4.5

- Object : 객체 데이터 // key & value로 구성되어있는 자료형
  (*address: {city : "Seoul", country : "South Korea"}
- Null : Null 데이터 저장할 수 있는 데이터 형태
  (*Null => 결측치)
- ObjectId : 각각의 문서 ID를 저장하는데 사용되는 타입
- Date : 날짜 데이터 및 시간 정보를 저장하는데 사용되는 타입

````python
#### 7.논리연산자
- $or : or 조건
- $and : and 조건
- $not
#### 8. MongoDB Aggregation Framework
- MongoDB > NoSQL , Aggregation => 집합|집계
- Aggregation => 프레임워늨
- $push <-> 배열이 아닌 요소들을 하나의 배열 자료 구조 형태로 만들어주는 기능
- 프레임워크 | 라이브러리
- 기존에 학습했던 find()의 상위호환 버전이라고 생각해도 무방
- Linux 운영체제 => pipeline 개념을 벤치마킹 => 함수의 기능이 구현 => 기능
- shard 샤드 => A (*1번 샤드) - > B(*2번 샤드)
- Aggregation Framework을 실행하기 위한 주요 단계
  - \$addFields: 기존 필드에 추가적으로 속성 더하거나 변경하고자 할 떄
  - \$match : 조건에 맞는 문서를 필터링
  - \$group : 문서를 그룹화하고 집계 작업을 수행할 수 있도록 준비
  - \$sort : 연산처리된 결과를 정렬
  - \$project : 연산처리된 결과를 특정 요소에 포함시키거나 새로운 문서로 변환 작업
  - \$limit : 특정 개수의 문서를 출력하고자 할 때
  - \$sort : 연산처리된 결과를 정렬
  - \$count : 연산결과값을 통해서 받아온 문서 수의 개수를 집계
  - \$lookup : 서로 다른 컬렉션을 묶어서 편집할 때 사용
     (* 연산처리를 하기 위한 집계 작접 명령어 = 예약어 )
    - \$sum : 특정 필드명 안에 포함되어있는 값의 합계
    - \$avg : 특정 필드명 안에 포함되어있는 값의 평균
    - \$min : 특정 필드명 안에 포함되어있는 최소값
    - \$max :특정 필드명 안에 포함되어있는 최대값
    - \$first : 특정 필드 안에 마지막번째 값
    - \$last : 특정 필드 안에 첫 번째 값
    - \$push : 기존값들을 배열의 형태로 구성할 떄 사용
    - \$convert : 특정 필드 내 속성을 변화시킬 때 사용
    - \$strLenCP : 문자열의 길이값을 찾아올 때 사용
    - \$toString : 문자열 형태가 아닌 타입의 값을 문자열로 변환
    - \$count : 연산 결과값을 통해서 받아온 문서 수의 개수를 집계
    - \$addToSet : 특정 필드 안에 있는 값을 중복하지 않고 찾아오고자 할 때
    - \$concat : 각 필드 안에 존재하는 값을 하나의 문자열로 생성하고자 할 때
    - \%documents : 임시 메모리 공간에 문서를 생성해놓을 때 사용
    - \$merge : 임시 메모리 공간에 문서를 특정 컬렉션 안에 삽입
    - \$facet: 어떤 하나의 측면 => interface ; aggregation 프레임워크를 사용할 때, 단점을 보완할 수 있음 => 파이프라인 => 앞에서 설정해놓은 단계가 종료되어야 그 다음 단계가 실행
      (* 만약, match연산을 통해서 찾아와야하는 조건식이 100줄 // group 3줄 => 파이프라인 내 상호 간 순서에 영향을 받지 않는 요소라고 한다면, 동시 실행을 시킬 수 있음 , 앞쪽에서 실행되어야할 요소를 기다리지 않고 병렬로 같이 실행 => 그룹을 생성 A그룹 & B그룹
    - \$redact : 사전적 의미 => "수정하다" (* 어떤 조건에 따라서 수정 여부를 판단)
      \$cond : 조건을 따져오는 역할
* 필드명 작성 순서
1. $match, $group : 해당 명령어 안에 어떤 조건식을 통해서 결과값이 도출되었음. 그리고 해당 결과값을 어떤 필드에 값을 담아야 하는 경우 => 무조건 내부적으로 필드명을 먼저 사용
2. $count, $sum, $avg : 어떤 값들을 새롭게 집계한 후 새로운 필드를 생성 할 때, 이때에는 무조건 예약어 를 먼저 사용 후 이후에 속성값으로 필드명을 입력
#### 9. SQL JOIN VS NoSQL $lookup
1) SQL => RDBMS (*관계형 데이터베이스 관리 시스템)
   - schema 개념 // 테이블을 생성, 필드릉 선언 => Primary key, Foreign key => index
   - INNER JOIN : 서로 다른 테이블에서 교집합 (*매칭되는 행만 가져옴)
   - LEFT OUTER JOIN : 메인이 되는 왼쪽 테이블을 모두 가져오고, 서브가 되는 오른쪽 테이블은 매칭 되는 값만 가져옴
   - RIGHT OUTER JOIN : LEFT OUTER JOIN 의 반대개념
2) NoSQL => 기본적으로 무조건 LEFT OUTER JOIN 의 형태를 가지고 있다.
3)
```
db.comments.aggregate([
    {
        $lookup:
            {
                from :"movies" // 조인할 대상의 컬렉션 이름
                localField: "movie_id"// 현재 컬렉션에서 조인에 사용할 필드
                foreignField: "_id" // 조인할 컬렉션에서 조인에 사용할 필드
                as: "movie"// 조인된 결과를 저장할 필드 이름
            }
    }
])
#### $$ 가 들어간 사용자 정의 변수 = 시스템 변수
- \$$ROOT : 현제 루트 폴더 문서 전체를 찾아오고자 할 때
- \$$NOW : 현재 날짜, 시간 조회
- \$$THIS : 배열 순회 중 현재 요소
````
