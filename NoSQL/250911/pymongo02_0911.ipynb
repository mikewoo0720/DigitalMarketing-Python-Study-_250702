```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

db = client["sample_mflix"]

users = db.users
comments = db.comments
movies = db.movies

print("ok", db.name, "collections :", [c for c in db.list_collection_names() if c in {"users", "comments", "movies"}])

print(db)
```

```python
# "Star"라는 단어가 포함된 영화 제목을 가져오세요!

for movie in movies.find({"title": {"$regex": "Star", "$options": "i"}},{"_id": 0, "title": 1}).limit(10) :
    print(movie)
```

```python
# "Star"라는 단어가 포함된 영화 제목을 가져오세요!

star_movies = movies.find({"title": {"$regex": ".*Star.*"}}, {"title": 1, "_id": 0})

for movie in star_movies :
    print("영화제목 :", movie["title"])
```

```python
# 각 감독들이 제작한 영화의 수가 25개 이상인 감독들만 찾아서 출력해주세요.

pipeline = [
    {"$unwind": "$directors"},
    {"$group": {
        "_id": "$directors",
        "count": {"$sum": 1}
    }},
    {"$match": {"count": {"$gte": 25}}},
    {"$sort": {"count": -1}},
    # {"$limit": 5}
]

for movie in movies.aggregate(pipeline):
    print(f"감독이름 : {movie["_id"]} / 영화 수 : {movie["count"]}")
```

```python
# 각 감독들이 제작한 영화의 수가 25개 이상인 감독들만 찾아서 출력해주세요.

director_Pipelines = [
    {"$unwind": "$directors"},
    {"$group": {"_id": "$directors", "count": {"$sum": 1}}},
    {"$match": {"count": {"$gte": 25}}}
]

diretor_counts = list(movies.aggregate(director_Pipelines))

for director in diretor_counts :
    print(f"감독 : {director["_id"]}, 영화 수 : {director["count"]}")
```

```python
# 평점을 기준으로 상위 5개의 영화를 찾으세요! (단, 투표를 1000표 이상 받은 영화에 한함)

pipeline = [
    {"$match": {"imdb.votes": {"$gte": 1000}}},
    {"$sort": {"imdb.rating": -1}},
    {"$limit": 5},
    {"$project": {
        "_id": 0,
        "title": 1,
        "rating": "$imdb.rating",
        "votes": "$imdb.votes"
    }}
]

for movie in movies.aggregate(pipeline):
    print(movie)
```

```python
# 평점을 기준으로 상위 5개의 영화를 찾으세요! (단, 투표를 1000표 이상 받은 영화에 한함)

top_rated_movies = list(movies.find({"imdb.votes": {"$gte": 1000}}).sort("imdb.rating", -1).limit(5))

for movie in top_rated_movies :
    print(f"영화제목 : {movie["title"]}, 평점 : {movie["imdb"]["rating"]}")
```
