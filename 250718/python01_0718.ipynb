현재 공유받은 크롤링 증권 사이트 안에 회사명만 한줄씩 출력해주세요

```python
import requests
from bs4 import BeautifulSoup

res = requests.get("https://davelee-fun.github.io/blog/crawling_stock_example.html")
soup = BeautifulSoup(res.content, "html.parser")
                     
datas = soup.select("ul.table-stock.sty1>li.row_sty")

for data in datas :
    name = data.select_one("div.st_name").get_text().strip().replace(" ", "").replace("\n", "")
    price = data.select_one("div.st_price").get_text().strip().replace(" ", "")
    rate = data.select_one("div.st_rate").get_text().strip().replace("\n", "").replace(" ", "")
    print(f"이름: {name:<15}, 주식: {price:<10}, 변동률: {rate:<10}")
```
